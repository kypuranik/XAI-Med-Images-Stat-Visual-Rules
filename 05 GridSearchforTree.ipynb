```python
import json
import os

# Final Decision Tree configuration for BTTypes
dt_params = {
    'criterion': 'entropy',
    'max_depth': 5,
    'min_samples_leaf': 4,
    'min_samples_split': 2
}

stat_feature_dir = 'zfmis_selected_features_BTTypes/k_3'

best_dt_params_path = os.path.join(
    stat_feature_dir,
    'best_dt_params.json'
)

with open(best_dt_params_path, 'w') as file:
    json.dump(
        {
            'train_file': os.path.join(
                stat_feature_dir,
                'train_selected_features.csv'
            ),
            'best_dt_params': dt_params
        },
        file,
        indent=2
    )

print("Final BTTypes Decision Tree parameters:")
print(dt_params)
print(f"Saved to: {best_dt_params_path}")
```
