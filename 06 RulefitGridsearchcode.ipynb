```python
import json
import os

# Final RuleFit configuration for BTTypes
rf_params = {
    'tree_size': 3,
    'sample_fract': 0.7,
    'max_rules': 2000
}

output_dir = 'rulefit_BTTypes'
os.makedirs(output_dir, exist_ok=True)

params_path = os.path.join(output_dir, 'best_rf_params.json')

with open(params_path, 'w') as file:
    json.dump(
        {
            'best_rf_params': rf_params
        },
        file,
        indent=2
    )

print("Final BTTypes RuleFit parameters:")
print(rf_params)
print(f"Saved to: {params_path}")
```
